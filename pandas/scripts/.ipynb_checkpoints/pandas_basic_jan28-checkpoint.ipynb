# Basic Inspection (Easy)

```python
import pandas as pd

df = pd.read_csv('/home/vijay/Downloads/hr_data.csv')

print(df.head(5))

print(80 * '*')
# What's the shape and data types of hr_df?

print(df.shape)
print(df.dtypes)
print(80 * '*')
# Show summary stats for numeric columns.

print(df.describe())

# How many employees per department?
print(df['Department'].value_counts())
```

# Selection & Filtering (Medium)

```python
# Select top 5 highest salaries with names.

print(df['Salary'].nlargest(5))
df1 = df.nlargest(5, "Salary")[['Name', 'Salary']]
print(df1)

print(80 * '*')
# Filter IT employees with performance > 3.0.

filter_it = df[(df['Department'] == 'IT') & (df['Performance_Score'] > 3.0)]
print(filter_it)

print(80 * '*')
# Get employees joined before 2022.
df['Join_Date'] = pd.to_datetime(df['Join_Date'])
before_2022 = df[df['Join_Date'].dt.year < 2022]

print(before_2022)
```

# Grouping & Aggregation (Medium)

```python
# Average salary by department?

df_avg = df.groupby('Department')['Salary'].mean()
print(df_avg)

print('*' * 80)

# Count, mean, max salary per dept/gender.
data_agg = df.groupby(['Department', 'Gender'])['Salary'].agg(['count', 'mean', 'max'])
print(data_agg)

print('*' * 80)
#Top performing dept (mean score)?

dept_high_mean = df.groupby(['Department'])['Performance_Score'].mean().idxmax()
print(dept_high_mean)
```

# Manipulation & Cleaning (Medium-Hard)

```python
from datetime import date
# Add tenure column in years.
df['Tenure'] = (pd.to_datetime(date.today()) -  df['Join_Date']).dt.days/365.25
print(df)

print(80 * "*")
# Create salary buckets (low/medium/high).
df['salary_bucket'] = pd.cut(df['Salary'], bins=3, labels=['Low', 'Medium', 'High'])
print(df)

print(80 * "*")
# Rank salaries within each department.
df['Salary_Rank'] = df.groupby('Department')['Salary'].rank(ascending=False)
print(df)
```
